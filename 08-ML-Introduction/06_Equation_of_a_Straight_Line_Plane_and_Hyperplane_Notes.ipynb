# Equation of a Straight Line, Plane, and Hyperplane

These equations are important in machine learning because models such as linear regression, logistic regression, and support vector machines use similar ideas.

## 1. Straight line in 2D
A straight line on an `x-y` graph is commonly written as:

$$y = mx + c$$

- `m` is the **slope**: how much `y` changes when `x` increases by 1.
- `c` is the **y-intercept**: where the line meets the y-axis (`x = 0`).

**Example:** `y = 2x + 3` has slope `2` and meets the y-axis at `3`. When `x = 1`, `y = 5`.

## 2. Another form of a line
The same line can also be written as:

$$ax + by + c = 0$$

If `b ≠ 0`, rearrange it to get:

$$y = -\frac{a}{b}x - \frac{c}{b}$$

So the slope is `-a/b` and the intercept is `-c/b`.

**Example:** `2x - y + 4 = 0` becomes `y = 2x + 4`.

## 3. Machine-learning notation
When data has features, we usually write a 2D line as:

$$w_1x_1 + w_2x_2 + b = 0$$

- `x₁`, `x₂` are input features.
- `w₁`, `w₂` are weights (coefficients).
- `b` is the bias or intercept.

In short vector form:

$$w^T x + b = 0$$

Here, `wᵀx` means the dot product: `w₁x₁ + w₂x₂`.

## 4. Plane and hyperplane
- In **2 dimensions**, the equation describes a **line**.
- In **3 dimensions**, it describes a **plane**: `w₁x₁ + w₂x₂ + w₃x₃ + b = 0`.
- In **n dimensions**, it describes a **hyperplane**.

The general form is always:

$$w^T x + b = 0$$

A hyperplane has one dimension less than its input space. For example, a plane is a 2D surface inside 3D space.

## 5. Normal vector and the origin
The weight vector `w` is called the **normal vector**. It is **perpendicular** to the line, plane, or hyperplane.

If `b = 0`, the line or hyperplane passes through the origin. Its equation becomes:

$$w^T x = 0$$

For a non-zero point vector `x` on this origin-centred hyperplane, `wᵀx = 0` means `w` and `x` are perpendicular. This follows from the dot-product rule:

$$w^T x = ||w|| ||x|| cos(θ)$$

The value is zero when the angle `θ` is 90°.

## Why this matters in ML
A classification model can use `wᵀx + b = 0` as a **decision boundary**. The hyperplane separates data into two sides:

- `wᵀx + b > 0` → one class
- `wᵀx + b < 0` → the other class
- `wᵀx + b = 0` → exactly on the boundary

**Memory trick:** 2D → line, 3D → plane, many dimensions → hyperplane.
