# Distance of a Point from a Plane or Hyperplane

This idea is useful in classification models such as logistic regression and support vector machines (SVM). A model uses a line or hyperplane to separate classes.

## Hyperplane equation
A hyperplane is written as:

$$w^T x + b = 0$$

- `x` is a point (feature vector).
- `w` is the normal vector, so it is perpendicular to the hyperplane.
- `b` is the bias/intercept.

If `b = 0`, the hyperplane passes through the origin: `wᵀx = 0`.

## Distance formula
For a point `s`, the shortest distance to the hyperplane is measured along the perpendicular direction.

For `wᵀx + b = 0`:

$$distance = |w^T s + b| / ||w||$$

Here `||w||` means the length (magnitude) of `w`. The absolute-value bars make the ordinary distance always zero or positive.

If the hyperplane passes through the origin (`b = 0`), the formula becomes `|wᵀs| / ||w||`.

## Signed distance: which side is the point on?
To know the side of the hyperplane, use the **signed distance**:

$$signed-distance = (w^T s + b) / ||w||$$

- Positive value: point is on the side toward which `w` points.
- Negative value: point is on the opposite side.
- Zero: point lies exactly on the hyperplane.

A negative signed distance does **not** mean a negative physical distance. It only records the side.

## Simple 2D example
Consider the line:

$$2x_1 + x_2 - 4 = 0$$

Here `w = [2, 1]`, `b = -4`, and `||w|| = sqrt(2² + 1²) = sqrt(5)`.

For point `s = [3, 1]`:

$$signed-distance = (2(3) + 1 - 4) / sqrt(5) = 3 / sqrt(5)$$

The result is positive, so the point is on the positive side of the line. Its actual distance is also `3 / sqrt(5)`.

## Why it matters in ML
A classifier can use the sign of `wᵀx + b` to assign a class:

- `wᵀx + b > 0` → one class
- `wᵀx + b < 0` → the other class

SVM also cares about the distance: it aims to place its boundary with a large gap (margin) between the classes.

**Memory trick:** absolute value gives the distance; the sign gives the side.
