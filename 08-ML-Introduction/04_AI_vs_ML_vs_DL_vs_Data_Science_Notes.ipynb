# AI vs Machine Learning vs Deep Learning vs Data Science

## Quick idea
These terms are connected, but they are not the same. Think of them like this:

```
Artificial Intelligence (AI)
└── Machine Learning (ML)
    └── Deep Learning (DL)

Data Science overlaps with all of them.
```

## 1. Artificial Intelligence (AI)
AI is the broad field of making computers perform tasks that normally need human intelligence, such as understanding language, recognising images, making decisions, or planning.

AI systems can work automatically, but some can also involve human review.

**Simple examples**
- Netflix suggests movies you may like.
- A self-driving car detects roads, traffic lights, and other vehicles.
- A chatbot answers customer questions.

## 2. Machine Learning (ML)
ML is a subset of AI. Instead of writing every rule by hand, we give a computer data and let it learn useful patterns.

ML is commonly used to analyse data, make predictions, classify items, and forecast future values.

**Simple example**
Give a model old house prices, location, size, and number of rooms. It learns patterns from the old data and predicts the price of a new house.

## 3. Deep Learning (DL)
DL is a subset of ML that uses neural networks with many layers. These networks are inspired by the way the brain processes information, but they are much simpler than a real brain.

Deep learning is especially useful for complex data such as images, speech, video, and text. It often needs a large amount of data and computing power.

**Simple examples**
- Face unlock on a phone.
- Speech-to-text in a voice assistant.
- Identifying a cat or dog in a photo.

## 4. Data Science
Data science is the work of using data to answer questions and solve real problems. It combines maths, statistics, programming, data cleaning, visualisation, and sometimes ML or DL.

A data scientist may do exploratory data analysis (EDA), build a prediction model, or explain results to a business team. Therefore, data science overlaps with AI, ML, and DL instead of sitting neatly inside one of them.

**Simple example**
A shop wants to know why sales dropped. A data scientist cleans sales data, creates charts, finds the reason, and may build a model to forecast next month's sales.

## Comparison at a glance

| Term | Main focus | Example |
|---|---|---|
| AI | Making machines act intelligently | Chatbot or self-driving car |
| ML | Learning patterns from data | Spam-email prediction |
| DL | Learning with multi-layer neural networks | Face recognition |
| Data Science | Getting useful answers from data | Sales analysis and forecasting |

## Key takeaway
- **AI** is the biggest umbrella.
- **ML** is one way to build AI by learning from data.
- **DL** is a specialised part of ML that uses deep neural networks.
- **Data Science** uses data, statistics, programming, and sometimes AI/ML/DL to solve problems.

**Memory trick:** AI is the goal, ML is a learning method, DL is a powerful type of ML, and Data Science is the broader problem-solving practice around data.
