# AI vs ML vs DL vs Data Science

## Quick revision notes
These names are connected. Do not worry about big words—think of them as different jobs done with computers and data.

## The picture to remember

```
Artificial Intelligence (AI)
└── Machine Learning (ML)
    └── Deep Learning (DL)

Data Science works across all of them.
```

Think of AI as the biggest box, ML as a smaller box inside it, and DL as an even smaller box inside ML.

## AI: make a computer act smart
AI means making a computer do useful tasks that normally need human thinking.

**Example: a movie app**
The app suggests action movies after you watch action movies. In simple English, it tries to guess what you may like next.

## ML: let the computer learn from examples
ML is one way to build AI. Instead of writing every rule by hand, we show the computer old examples and let it find a pattern.

**Example: spam email**
Show the computer old emails marked `spam` and `not spam`. It learns common spam patterns, then labels a new email.

## DL: ML for hard things like photos and speech
DL uses neural networks with many layers. It is useful when the data is big and complicated.

**Example: face unlock**
The phone learns face patterns from many pictures. When it sees your face, it checks whether the pattern looks like you.

## Data Science: use data to answer questions
Data Science includes collecting data, cleaning it, making charts, finding answers, and sometimes building ML models.

**Example: falling sales**
A data scientist checks sales charts and finds that a product was out of stock. This is useful data science even without ML.

| Term | Tiny memory line |
|---|---|
| AI | Make computers act smart |
| ML | Learn from old data |
| DL | Learn complex patterns with many layers |
| Data Science | Use data to understand and solve problems |

**Memory trick:** `AI is the big umbrella; ML learns; DL learns deeply; Data Science explains data.`
