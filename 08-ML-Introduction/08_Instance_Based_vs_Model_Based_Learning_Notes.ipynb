# Instance-Based vs Model-Based Learning

Machine-learning systems can make predictions in two broad ways: by comparing a new item with stored examples, or by first learning a general pattern from the data.

## 1. Instance-based learning
An instance-based method keeps the training examples. When a new input arrives, it compares that input with nearby or similar examples and then makes a prediction.

It is often called a **lazy learner** because most of the work happens when a prediction is requested, not during an earlier training step.

**Example: K-Nearest Neighbours (KNN)**
To predict whether a student will pass or fail, KNN finds students with similar study hours and play hours. If most nearby students passed, it predicts `pass`.

## 2. Model-based learning
A model-based method learns a pattern, rule, equation, or decision boundary from the training data before it receives new inputs. It then uses the learned model to predict.

**Example**
For the same student data, a model may learn a decision boundary: students on one side are likely to pass and students on the other side are likely to fail. A new student is classified using that learned boundary.

Examples include linear regression, logistic regression, decision trees, and neural networks.

## Key difference
- **Instance-based learning:** “Which stored examples are most similar to this new one?”
- **Model-based learning:** “What general pattern did I learn from the training data?”

Instance-based methods rely directly on stored examples. Model-based methods try to generalise from examples, so they can make predictions using learned parameters or rules.

## Comparison at a glance

| Point | Instance-based | Model-based |
|---|---|---|
| Main idea | Compare with stored training examples | Learn a pattern or model |
| Training time | Usually low | Usually higher |
| Prediction time | Can be slower for large datasets | Often faster after training |
| Training data at prediction time | Usually needed | Often not needed after training |
| Storage | Can be high because examples are stored | Often lower, but depends on model size |
| Example | KNN | Linear/logistic regression, decision tree |

## Simple analogy
- **Instance-based:** Before answering a question, look through many old solved examples and choose the most similar ones.
- **Model-based:** Study the old examples first, learn the rule, and then answer new questions quickly using that rule.

Neither approach is always best. The right choice depends on the data, the task, available storage, and how quickly predictions are needed.

## Key takeaway
- Instance-based learning predicts from similar stored examples.
- Model-based learning predicts from a learned general pattern.
- KNN is a common instance-based algorithm.

**Memory trick:** `Instances = compare examples`; `Models = learn rules`.
