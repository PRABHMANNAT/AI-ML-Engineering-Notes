# Instance-Based vs Model-Based Learning

## Easy, quick revision notes
Both methods use old examples to answer a new question. The big difference is: **does the machine look at old examples now, or did it learn a rule earlier?**

## The easiest idea
- **Instance-based learning:** keep old examples. For a new example, look nearby and copy what most similar examples did.
- **Model-based learning:** study old examples first. Learn one useful rule, then use that rule for new examples.

**Like you are 5:** instance-based asks, “Which old toys look most like this toy?” Model-based says, “I learned the sorting rule, so I can use the rule.”

## Example: will a student pass?
We use two inputs: `play_hours` and `study_hours`. The answer is `Pass` or `Fail`.

Usually, more study and less play makes passing more likely. Some people are different, so a few dots may not follow the usual pattern. That is normal.

```python
import numpy as np
import matplotlib.pyplot as plt

# Each old student is one dot. 1 means Pass; 0 means Fail.
play_hours = np.array([1, 2, 2, 3, 4, 5, 6, 7, 8])
study_hours = np.array([7, 6, 5, 5, 4, 3, 3, 2, 1])
passed = np.array([1, 1, 1, 1, 1, 0, 0, 0, 0])
colors = np.where(passed == 1, '#2a9d8f', '#f4a261')
new_student = np.array([4.0, 4.0])
old_students = np.column_stack([play_hours, study_hours])

# Instance-based method: find the 3 closest old students.
distances = np.linalg.norm(old_students - new_student, axis=1)
nearest = np.argsort(distances)[:3]

fig, axes = plt.subplots(1, 2, figsize=(13, 5), sharex=True, sharey=True)
for ax in axes:
    ax.scatter(play_hours, study_hours, c=colors, s=75, label='Old students', zorder=2)
    ax.scatter(*new_student, marker='*', s=220, color='#d62828', label='New student', zorder=3)
    ax.set_xlabel('Play hours')
    ax.grid(alpha=0.25)
axes[0].set_ylabel('Study hours')

# Left: point to nearby examples.
for index in nearest:
    axes[0].plot([new_student[0], play_hours[index]], [new_student[1], study_hours[index]],
                 '--', color='#d62828', alpha=0.8)
axes[0].set_title('Instance-based: ask nearby dots', weight='bold')
axes[0].text(0.03, 0.04, 'The 3 nearest old students vote.', transform=axes[0].transAxes,
             bbox=dict(boxstyle='round', facecolor='white', alpha=0.9))

# Right: show one simple learned-style boundary.
boundary_x = np.linspace(0, 9, 100)
axes[1].plot(boundary_x, boundary_x - 1, color='#264653', linewidth=2.4, label='Learned-style rule')
axes[1].text(0.5, 6.8, 'Pass side', color='#2a9d8f', weight='bold')
axes[1].text(5.6, 1.0, 'Fail side', color='#d62828', weight='bold')
axes[1].set_title('Model-based: use one learned rule', weight='bold')

axes[0].legend(loc='upper right', fontsize=8)
axes[1].legend(loc='upper right', fontsize=8)
plt.suptitle('Two ways to answer the same new question', weight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

**What the diagram shows:** each dot is one old student. Green means Pass; orange means Fail; the red star is a new student.

- Left: instance-based learning checks nearby dots.
- Right: model-based learning uses one boundary/rule. The line is only a simple picture of a rule that a real model learns from data.

## 1. Instance-based learning = “look nearby”
The method keeps the old examples. When a new student comes, it finds the most similar old students and lets them vote.

The most common example is **K-Nearest Neighbours (KNN)**. `K` means how many nearby examples we ask. If `K = 3`, we ask the 3 closest examples.

```python
old_students = np.column_stack([play_hours, study_hours])
new_student = np.array([4.0, 4.0])
k = 3
distances = np.linalg.norm(old_students - new_student, axis=1)
nearest = np.argsort(distances)[:k]
vote = passed[nearest].mean()
print('Nearest students:', nearest)
print('Prediction:', 'Pass' if vote >= 0.5 else 'Fail')
```

**What the code does:** it finds the three closest old students. If most of them passed, it predicts `Pass`; otherwise it predicts `Fail`.

## 2. Model-based learning = “learn a rule”
A model-based method studies the old dots **before** a new question arrives. It learns a general rule, such as a line, curve, tree, or decision boundary.

For a new student, it uses that saved rule. It usually does not need to search through every old dot each time.

Examples: Linear Regression, Logistic Regression, Decision Trees, and Neural Networks.

```python
def simple_rule(play, study):
    return study - play + 1

score = simple_rule(*new_student)
print('Rule score:', score)
print('Prediction:', 'Pass' if score >= 0 else 'Fail')
```

**What the code does:** it uses one learned-style rule instead of searching through all old students. The rule gives a score, then turns the score into `Pass` or `Fail`.

## Quick comparison

| Question | Instance-based | Model-based |
|---|---|---|
| What does it do first? | Keeps examples | Learns a rule from examples |
| Example | KNN | Logistic Regression |
| What is commonly saved? | Many/all training examples | Learned weights, tree, or rule |
| Prediction speed | Can be slower with lots of data | Often faster after training |
| Best simple memory | “Look nearby” | “Use the rule” |

## When to use each one
- Use **instance-based** learning when “similar things behave similarly” and keeping/searching old examples is okay.
- Use **model-based** learning when you want one learned rule and usually faster predictions.

**Important:** a learned rule is not magic. It can still make mistakes if the training data is poor or the rule is too simple/too complicated.

**Memory trick:** KNN keeps examples. A model learns a rule.
