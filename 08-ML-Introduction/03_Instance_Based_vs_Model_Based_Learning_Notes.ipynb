# Instance-Based vs Model-Based Learning

## Quick revision notes
Both methods use old examples to answer a new question. The big difference is **how** they remember those examples.

## The easiest idea
- **Instance-based learning:** look at similar old examples, then decide.
- **Model-based learning:** learn a rule first, then use the rule to decide.

**Like a child:** instance-based is looking at old homework before answering. Model-based is learning the rule and answering without opening the old homework.

## Example: will a student pass?
We use two inputs: `play_hours` and `study_hours`. The answer is `Pass` or `Fail`.

Usually, more study and less play makes passing more likely. Real data can have exceptions, and that is normal.

```python
import numpy as np
import matplotlib.pyplot as plt

play_hours = np.array([1, 2, 2, 3, 4, 5, 6, 7, 8])
study_hours = np.array([7, 6, 5, 5, 4, 3, 3, 2, 1])
passed = np.array([1, 1, 1, 1, 1, 0, 0, 0, 0])
colors = np.where(passed == 1, 'green', 'orange')
plt.scatter(play_hours, study_hours, c=colors, s=80)
plt.xlabel('Play hours'); plt.ylabel('Study hours')
plt.title('Green = Pass, Orange = Fail'); plt.grid(); plt.show()
```

**What the diagram shows:** each dot is one old student. Green dots passed and orange dots failed. A new student can be compared with these old dots.

## 1. Instance-based learning
The method stores the old examples. When a new student comes, it looks for the nearest old students and lets them vote.

The most common example is **K-Nearest Neighbours (KNN)**. `K` means how many nearby examples we ask.

```python
old_students = np.column_stack([play_hours, study_hours])
new_student = np.array([4.5, 3.5])
k = 3
distances = np.linalg.norm(old_students - new_student, axis=1)
nearest = np.argsort(distances)[:k]
vote = passed[nearest].mean()
print('Nearest students:', nearest)
print('Prediction:', 'Pass' if vote >= 0.5 else 'Fail')
```

**What the code does:** it finds the three closest old students. If most of them passed, it predicts `Pass`; otherwise it predicts `Fail`.

## 2. Model-based learning
A model-based method studies the old dots once and learns a general rule, such as a line or decision boundary. It does not need to search through every old dot for each new prediction.

Examples: Linear Regression, Logistic Regression, Decision Trees, and Neural Networks.

```python
def simple_rule(play, study):
    return study - play + 1

score = simple_rule(*new_student)
print('Rule score:', score)
print('Prediction:', 'Pass' if score >= 0 else 'Fail')
```

**What the code does:** it uses one learned-style rule instead of searching through all old students. The rule gives a score, then turns the score into `Pass` or `Fail`.

## Quick comparison

| Point | Instance-based | Model-based |
|---|---|---|
| Main idea | Compare new input with old examples | Use a learned rule |
| Example | KNN | Logistic Regression |
| Stores | Usually stores training examples | Stores learned weights/rules |
| Prediction speed | Can be slower with lots of data | Often faster after training |
| Best simple memory | “Look nearby” | “Use the rule” |

## When to use each one
- Use **instance-based** learning when similar nearby examples are useful and the dataset is not too large.
- Use **model-based** learning when you want a general rule and faster predictions.

**Memory trick:** KNN keeps examples. A model learns a rule.
