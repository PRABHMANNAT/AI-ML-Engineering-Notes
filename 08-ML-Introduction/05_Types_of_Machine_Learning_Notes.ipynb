# Types of Machine Learning

There are three common ways for a model to learn:

1. **Supervised learning** — learn from labelled examples.
2. **Unsupervised learning** — find hidden groups or patterns.
3. **Reinforcement learning** — learn by trying actions and receiving rewards.

## 1. Supervised learning
In supervised learning, the training data includes **inputs** and the correct **output (label)**. The model learns the relationship between them, then predicts the output for new inputs.

**House-price example**
- Inputs / independent features: house size, number of rooms, location.
- Output / dependent feature: house price.
- Goal: predict the price of a new house.

The word *supervised* means the model has known answers to learn from.

### Regression vs classification
Supervised problems are usually one of these two types:

| Type | Output | Simple example |
|---|---|---|
| **Regression** | A continuous number | Predict a house price or tomorrow's temperature |
| **Classification** | A category or label | Predict spam/not spam or pass/fail |

**Binary classification** has two classes, such as `pass` or `fail`.  
**Multiclass classification** has more than two classes, such as `low`, `medium`, or `high`.

## 2. Unsupervised learning
In unsupervised learning, the data has **no known output label**. The goal is not to predict a given answer; it is to discover useful patterns, groups, or structure in the data.

**Customer-segmentation example**
An online shop has customer salary and spending-score data. It can group customers into clusters such as:
- high salary, high spending
- low salary, low spending
- medium salary, high spending

The shop can then send a suitable offer to each group. Common clustering algorithms include **K-Means**, **Hierarchical Clustering**, and **DBSCAN**.

## 3. Reinforcement learning
In reinforcement learning, an **agent** learns by taking actions in an environment. Good actions receive rewards; bad actions may receive low rewards or penalties. Over time, the agent learns a better strategy.

**Simple example**
A game-playing AI tries different moves. Winning gives a high reward, so the AI gradually learns which moves are more useful.

Other examples: robot navigation, self-driving decisions, and game-playing systems.

## Quick comparison

| Learning type | What is given? | Main goal |
|---|---|---|
| Supervised | Inputs + correct labels | Predict an output |
| Unsupervised | Inputs only | Find patterns or clusters |
| Reinforcement | Actions, rewards, and feedback | Learn the best action strategy |

## Key takeaway
- Use **supervised learning** when you already know the answers in past data.
- Use **unsupervised learning** when you want to explore and group unlabeled data.
- Use **reinforcement learning** when an agent must improve through trial, feedback, and rewards.

**Memory trick:** `Labels → Supervised`, `Groups → Unsupervised`, `Rewards → Reinforcement`.
