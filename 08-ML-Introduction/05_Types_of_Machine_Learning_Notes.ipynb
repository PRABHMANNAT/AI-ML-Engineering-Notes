# Types of Machine Learning

## Quick revision notes
There are three main ways a computer can learn: learn from answers, find groups, or learn from rewards.

## The three types

| Type | Easy meaning | Example |
|---|---|---|
| **Supervised learning** | Learn from questions that already have answers | Predict a house price |
| **Unsupervised learning** | Find groups when no answers are given | Group similar customers |
| **Reinforcement learning** | Try actions and learn from rewards | Learn to play a game |

## 1. Supervised learning = learn with an answer sheet
The old data has inputs and the correct answer (label). The computer learns from those answers.

**Example: house price**
Old data says: a house has this size, this many rooms, and this price. The computer learns the pattern and guesses the price of a new house.

| Problem | What it predicts | Example |
|---|---|---|
| **Regression** | A number | Price = 500,000 |
| **Classification** | A name/category | Pass or fail |

`Pass/fail` is **binary** classification because there are two answers. `Low/medium/high` is **multiclass** classification because there are more than two answers.

## 2. Unsupervised learning = find friends in a crowd
There is no answer column. The computer looks at the data and makes groups of things that look alike.

**Example: customers**
A shop has income and spending data. The computer may group people who earn and spend a lot together. The shop can then give that group a suitable offer.

Common grouping methods: **K-Means**, **Hierarchical Clustering**, and **DBSCAN**.

## 3. Reinforcement learning = learn by trying
An **agent** tries an action. Good actions get rewards; bad actions get low rewards or penalties. With practice, the agent learns better actions.

**Example: a game**
A game AI tries a move. Winning gives it a reward, so it learns to use moves that help it win more often.

Simple words: `agent` = learner, `action` = choice, `reward` = feedback.

## One-minute revision
- Have old data with correct answers? → **Supervised**
- Have data but no answers, and need groups? → **Unsupervised**
- Have actions and rewards? → **Reinforcement**

**Memory trick:** `Answers → Supervised; Groups → Unsupervised; Rewards → Reinforcement.`
