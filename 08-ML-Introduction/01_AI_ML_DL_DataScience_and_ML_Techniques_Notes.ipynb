## The four terms in one table

| Term | Simple meaning | Example |

## Relationship to remember

```
AI
└── ML
    └── DL

Data Science overlaps with all three.
```

- **AI** is the biggest goal: make a system act intelligently.
- **ML** is one method for reaching that goal: learn from examples.
- **DL** is a specialised ML method for complex data such as images, text, and speech.
- **DS** uses data to understand a problem and support decisions.

## 1. Artificial Intelligence (AI)
AI is the broad field of building systems that can make decisions, understand information, recommend items, or take useful actions.

**Example: movie recommendation**
A streaming app suggests action movies after you watch several action movies. In simple English, the app uses your past behaviour to guess what you may enjoy next.

AI can work automatically, but some AI systems still need human checking or approval.

## 2. Machine Learning (ML)
ML learns patterns from old data instead of using only hand-written rules. It can then make a prediction for new data.

**Example: house-price prediction**
Give the model past house details such as size, rooms, and price. It learns that larger houses usually cost more. For a new house, it uses that learned pattern to estimate the price.

ML is often used for prediction, classification, forecasting, recommendations, and finding unusual activity.

## 3. Deep Learning (DL)
DL is a part of ML that uses neural networks with many layers. It is helpful when the data is complex, for example images, speech, video, or natural language.

**Example: face unlock**
The phone studies many face patterns and learns features such as the shape and arrangement of facial parts. When you look at the phone, it compares the new image with the learned pattern.

DL usually needs more data and computing power than simpler ML methods.

## 4. Data Science (DS)
Data Science is the full process of using data to answer questions and solve real problems. Typical work includes cleaning data, making charts, using statistics, building models, and explaining results.

**Example: sales drop**
A data scientist checks sales by month, product, and location. Charts may show that one product was unavailable in one city. This is useful even if no ML model is built.

# Three main types of Machine Learning

| Type | What data is given? | Main goal | Easy example |
|---|---|---|---|
| **Supervised** | Inputs and correct answers (labels) | Predict an answer | Predict house price |
| **Unsupervised** | Inputs only; no correct answer | Find hidden groups or patterns | Group similar customers |
| **Reinforcement** | Actions and rewards | Learn the best action | Learn to play a game |

## Supervised learning
In supervised learning, the dataset contains **input features** and a known **target/label**. The model learns from these known answers.

**Example: student result**
- Inputs: study hours and play hours
- Target: pass or fail

The model learns from past students. For a new student, it predicts `pass` or `fail` using the study and play pattern.

| Problem | Output type | Example |
|---|---|---|
| **Regression** | A meaningful number | House price, temperature, sales |
| **Classification** | A category | Pass/fail, spam/not spam |

**Binary classification** has two classes. **Multiclass classification** has more than two classes, such as low, medium, and high risk.

## Unsupervised learning
Unsupervised learning has no target column. The model looks for natural groups or patterns in the data.

**Example: customer groups**
An online shop has income and spending-score data. The model may create groups such as high-income/high-spending and low-income/low-spending customers. It is not predicting the spending score; it is finding similar customers.

Common clustering algorithms: **K-Means**, **Hierarchical Clustering**, and **DBSCAN**.

## Reinforcement learning
In reinforcement learning, an **agent** takes actions and receives rewards or penalties. It improves by trying, receiving feedback, and choosing better actions over time.

**Example: game-playing AI**
The AI tries a move. Winning gives a high reward, so the AI learns to choose moves that improve its chance of winning.

Key words: `agent` = learner, `action` = choice, `reward` = feedback, `environment` = world where the agent acts.

## Quick algorithm map

| Task | Common algorithms |
|---|---|
| Regression | Linear Regression, Ridge, Lasso, Elastic Net |
| Classification | Logistic Regression, Decision Tree, Random Forest, XGBoost |
| Both regression and classification | Decision Tree, Random Forest, boosting methods |
| Clustering | K-Means, Hierarchical Clustering, DBSCAN |

Start with the problem type, not the algorithm name. First decide whether you have labels, need groups, or need reward-based learning.

## Exam-ready summary
- **AI**: broad goal of intelligent computer systems.
- **ML**: AI method that learns patterns from data.
- **DL**: ML using multi-layer neural networks.
- **Data Science**: uses data, statistics, analysis, and sometimes ML to solve problems.
- **Supervised**: labels are available; predict an output.
- **Regression**: output is a number.
- **Classification**: output is a category.
- **Unsupervised**: no labels; find groups or patterns.
- **Reinforcement**: learn actions from rewards.

**Memory trick:** `AI is the umbrella; ML learns; DL learns deeply; DS explains data.`
