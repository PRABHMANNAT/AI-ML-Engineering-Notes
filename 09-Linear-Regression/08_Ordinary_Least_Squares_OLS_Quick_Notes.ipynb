# OLS: The Best-Fit Line, Made Simple

**OLS** means **Ordinary Least Squares**. It finds the straight line that makes the total squared prediction mistakes as small as possible.

**Tiny memory idea:** try to put one line as close as it can to all the dots.

## 1. What is the line for?

Imagine `x` is hours studied and `y` is a test score. A line lets us guess a score from study hours.

$$\hat{y} = \beta_0 + \beta_1x$$

- $x$: the thing we know (study hours).
- $\hat{y}$: the model's guess. The hat means “predicted.”
- $\beta_0$: where the line starts on the y-axis. It moves the line up or down.
- $\beta_1$: how much the guess changes when $x$ goes up by 1. It tilts the line.

Example: $\hat{y}=1+2x$ means each extra study hour adds 2 points to the model's guess.

## 2. What is a mistake (residual)?

A **residual** is the gap between the real answer and the line's guess:

$$\text{residual} = y - \hat{y}$$

If the real score is 7 and the line guesses 5, the residual is $7-5=2$. The dot is 2 points above the line.

A dot below the line has a negative residual. OLS squares each residual, so above-line and below-line mistakes cannot cancel each other. Squaring also makes a big miss count more than a small miss.

## 3. What does “least squares” mean?

Square every residual, then add them:

$$SSE = \sum_{i=1}^{n}(y_i-\hat{y}_i)^2$$

This total is called **SSE**: Sum of Squared Errors. OLS chooses the line with the **smallest SSE**.

Small example: residuals $+2$ and $-2$ add to zero, which hides the mistakes. Squared, they become $4+4=8$, so both misses count.

**OLS does not promise that the line touches every dot.** It finds the line whose squared misses add up to the smallest total.

## 4. How does OLS find the line?

For one input, OLS can calculate the slope and start point directly. It first finds the average $x$ and average $y$:

$$\beta_1 = \frac{\sum (x_i-\bar{x})(y_i-\bar{y})}{\sum (x_i-\bar{x})^2}$$
$$\beta_0 = \bar{y} - \beta_1\bar{x}$$

You can skip memorizing these long formulas for now. In plain English:

1. OLS checks how $x$ and $y$ move together to choose the tilt ($\beta_1$).
2. It moves the line up or down to pass through the middle of the data ($\beta_0$).
3. Together, these make the line with the smallest squared misses.

## 5. Tiny worked example

Suppose the data is:

| Study hours ($x$) | Score ($y$) |
|---:|---:|
| 1 | 2 |
| 2 | 3 |
| 3 | 5 |
| 4 | 4 |
| 5 | 6 |

OLS gives this line: $\hat{y}=1.3+0.9x$.

For 6 study hours, the guess is $1.3+0.9\times6=6.7$ points.

This is a prediction, not a promise. The real score can be different.

## 6. Diagram: dots, best-fit line, and the misses

Run the Python cell. The blue dots are real scores. The orange line is OLS's guess. Each green up-and-down mark shows one residual: the distance from a real dot to its guess on the line.

```python
import numpy as np
import matplotlib.pyplot as plt

# Five students: study hours and their real scores.
x = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([2, 3, 5, 4, 6], dtype=float)

# Find the OLS line with the simple one-input formulas.
x_average = x.mean()
y_average = y.mean()
slope = np.sum((x - x_average) * (y - y_average)) / np.sum((x - x_average) ** 2)
intercept = y_average - slope * x_average
guesses = intercept + slope * x

# Draw a smooth line across the graph.
x_line = np.linspace(0, 6, 100)
y_line = intercept + slope * x_line

fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(x, y, color='#1d3557', s=75, label='Real scores', zorder=3)
ax.plot(x_line, y_line, color='#f77f00', linewidth=2.5, label='OLS best-fit line')
for x_i, real_y, guessed_y in zip(x, y, guesses):
    ax.vlines(x_i, min(real_y, guessed_y), max(real_y, guessed_y),
              color='#2a9d8f', linewidth=2.5)
ax.set_xlabel('Study hours')
ax.set_ylabel('Score')
ax.set_title('OLS keeps the total squared gaps as small as possible')
ax.grid(alpha=0.25)
ax.legend()
plt.show()

print(f'Best-fit line: score = {intercept:.1f} + {slope:.1f} × study_hours')
print(f'Total squared error (SSE): {np.sum((y - guesses) ** 2):.2f}')
```

## 7. OLS and Gradient Descent

Both try to make squared mistakes small. They take different routes:

| OLS | Gradient Descent |
|---|---|
| Calculates the best line directly for ordinary linear regression. | Starts with a line and improves it in small steps. |
| No learning-rate choice is needed. | Needs a learning rate to choose step size. |
| A convenient fit for many standard datasets. | Useful when there are many parameters or the dataset is very large. |

A program usually uses dependable math tools to find the OLS answer; you do not need to calculate matrix inverses by hand.

## Quick revision

- **OLS** finds a straight line with the smallest total squared misses.
- A **residual** is the real answer minus the line's guess.
- **Least squares** means square each residual and add the squares (SSE).
- $\beta_0$ moves the line up/down; $\beta_1$ tilts it.
- OLS calculates the line directly; Gradient Descent improves a line step by step.

**Remember:** OLS picks the line that stays closest to all the dots overall.
