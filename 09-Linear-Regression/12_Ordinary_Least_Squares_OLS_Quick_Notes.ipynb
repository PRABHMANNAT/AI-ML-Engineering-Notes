# Ordinary Least Squares (OLS) - Quick Revision Notes

OLS is a way to find the best straight line for linear regression **directly using a formula**.

## 1. The line we want

$$\hat{y} = \beta_0 + \beta_1x$$

- $x$: input, for example years of experience.
- $\hat{y}$: predicted output, for example salary.
- $\beta_0$: intercept - where the line starts when $x=0$.
- $\beta_1$: slope/coefficient - how much the prediction changes when $x$ increases by 1.

Example: $\hat{y} = 1.3 + 0.9x$. If experience rises by 1 year, predicted salary rises by 0.9 units.

## 2. What does “least squares” mean?

For every data point, find the **residual** (the vertical gap):

$$residual_i = y_i - \hat{y}_i$$

OLS chooses $\beta_0$ and $\beta_1$ that make this total as small as possible:

$$SSE = \sum_{i=1}^{n}(y_i - (\beta_0 + \beta_1x_i))^2$$

The errors are squared, so positive and negative gaps cannot cancel out. Large mistakes also receive a larger penalty. Minimizing SSE and minimizing MSE give the same best line; MSE is just SSE divided by $n$.

## 3. OLS formulas for simple linear regression

First find the average input $\bar{x}$ and average output $\bar{y}$. Then:

$$\beta_1 = \frac{\sum(x_i-\bar{x})(y_i-\bar{y})}{\sum(x_i-\bar{x})^2}$$

$$\beta_0 = \bar{y} - \beta_1\bar{x}$$

Easy meaning:

- The slope asks: **when $x$ moves away from its average, does $y$ move with it?**
- The intercept then moves the line so it passes through the point $(\bar{x}, \bar{y})$.

## 4. OLS versus Gradient Descent

| OLS | Gradient Descent |
|---|---|
| Uses a direct mathematical solution for ordinary least squares. | Improves the coefficients step by step. |
| No learning-rate choice is needed. | Needs a learning rate and iterations. |
| Great for small or medium linear-regression problems. | Useful for very large datasets and many ML models. |

Both aim to minimize squared prediction errors. In practical software, OLS is usually solved with stable numerical methods rather than calculating a matrix inverse by hand.

## 5. Tiny worked example

Use $x = [1,2,3,4,5]$ and $y = [2,3,5,4,6]$.

- $\bar{x} = 3$ and $\bar{y} = 4$.
- OLS gives $\beta_1 = 0.9$ and $\beta_0 = 1.3$.
- Best-fit line: $\hat{y} = 1.3 + 0.9x$.
- For $x=6$, the prediction is $1.3 + 0.9(6) = 6.7$.

The data dots do not all sit exactly on the line. OLS finds the line whose **total squared vertical gaps** are smallest.

## Visual diagram

Run the next cell. Left: green lines are residuals from the OLS best-fit line. Right: each possible slope gives a total squared error; the lowest point is the OLS choice.

```python
import numpy as np
import matplotlib.pyplot as plt

# Small example dataset
x = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([2, 3, 5, 4, 6], dtype=float)

# OLS formulas for slope and intercept
x_mean = x.mean()
y_mean = y.mean()
slope = np.sum((x - x_mean) * (y - y_mean)) / np.sum((x - x_mean) ** 2)
intercept = y_mean - slope * x_mean
prediction = intercept + slope * x

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Left: fitted line and residuals
axes[0].scatter(x, y, color='#1d3557', s=65, label='Actual data', zorder=3)
axes[0].plot(x, prediction, color='#f77f00', linewidth=2.5, label='OLS best-fit line')
for xi, yi, y_hat in zip(x, y, prediction):
    axes[0].vlines(xi, min(yi, y_hat), max(yi, y_hat), color='#2a9d8f', linewidth=2)
axes[0].annotate('residual', xy=(3, prediction[2]), xytext=(3.35, 3.8),
                 arrowprops=dict(arrowstyle='->', color='#2a9d8f'), color='#137a6c', weight='bold')
axes[0].set_title('OLS chooses the line with small squared residuals', weight='bold')
axes[0].set_xlabel('Input x')
axes[0].set_ylabel('Output y')
axes[0].legend()
axes[0].grid(alpha=0.2)

# Right: show the SSE for many candidate slopes
candidate_slopes = np.linspace(-0.2, 2.0, 200)
sse_values = []
for candidate_slope in candidate_slopes:
    candidate_intercept = y_mean - candidate_slope * x_mean
    candidate_prediction = candidate_intercept + candidate_slope * x
    sse_values.append(np.sum((y - candidate_prediction) ** 2))
axes[1].plot(candidate_slopes, sse_values, color='#6a4c93', linewidth=2.5)
best_sse = np.sum((y - prediction) ** 2)
axes[1].scatter(slope, best_sse, color='#d62828', s=70, zorder=3, label='Smallest SSE')
axes[1].annotate('OLS slope', xy=(slope, best_sse), xytext=(1.2, best_sse + 5),
                 arrowprops=dict(arrowstyle='->', color='#d62828'), color='#d62828', weight='bold')
axes[1].set_title('The best slope sits at the bottom', weight='bold')
axes[1].set_xlabel('Candidate slope')
axes[1].set_ylabel('Sum of Squared Errors (SSE)')
axes[1].legend()
axes[1].grid(alpha=0.2)

plt.tight_layout()
plt.show()
print(f'OLS slope (beta1): {slope:.2f}')
print(f'OLS intercept (beta0): {intercept:.2f}')
```

## 6. Exam-ready recap

- OLS means Ordinary Least Squares.
- It finds the line that minimizes the total squared residuals (SSE).
- Simple regression uses $\beta_1 = \frac{\sum(x_i-\bar{x})(y_i-\bar{y})}{\sum(x_i-\bar{x})^2}$ and $\beta_0 = \bar{y}-\beta_1\bar{x}$.
- OLS is a direct formula-based method; Gradient Descent is an iterative method.

**One-line answer:** OLS finds the regression line whose total squared vertical distance from the data points is the smallest.
