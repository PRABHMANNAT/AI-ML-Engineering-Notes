# Overfitting, Underfitting, and Data Splits - Quick Revision Notes

**Main goal:** make a model that works not only on old training examples, but also on new unseen examples. This is called **generalization**.

## 1. Training, validation, and test data

Imagine 1,000 house records. We split them so each part has one job:

| Part | What it is for |
|---|---|
| **Training set** | The model learns patterns from this data. |
| **Validation set** | We choose settings such as model complexity or regularization using this data. |
| **Test set** | Final honest check on new data. Do not use it while choosing the model. |

A 70/30 train-test split is one possible start, not a fixed rule. Often, the training part is split again into train and validation, or cross-validation is used.

## 2. Underfitting

An **underfit** model is too simple to learn the real pattern. It is like trying to draw a curved road with one straight line.

- Training performance: poor.
- Validation/test performance: poor.
- Usual name: **high bias** (the model makes too many simple assumptions).
- Typical fixes: use better features, choose a more flexible model, or reduce too-strong regularization.

## 3. Good fit / generalized model

A good model learns the useful pattern without memorizing tiny random details.

- Training performance: good.
- Validation/test performance: also good and reasonably close to training performance.
- This is what we want: low error on both known and new data.

## 4. Overfitting

An **overfit** model memorizes the training data, including noise. It looks clever on old questions but struggles with new ones.

- Training performance: very good.
- Validation/test performance: much worse.
- Usual name: **high variance** (small changes in data can change the learned model a lot).
- Typical fixes: get more data, simplify the model, use regularization, select fewer features, or use early stopping.

## 5. Bias and variance in one table

| Model behavior | Train error | Test error | Usual description |
|---|---:|---:|---|
| Underfitting | High | High | High bias, usually low variance |
| Good fit | Low | Low and close to train error | Balanced bias and variance |
| Overfitting | Very low | High | Low bias, high variance |

**Quick rule:** a big gap between training and test performance is a warning sign of overfitting.

## Visual diagram

Run the next cell. All three panels see the same training dots. The left line misses the curved pattern (underfit); the middle curve captures the useful shape (good fit); the right curve bends too much to chase noise (overfit).

```python
import numpy as np
import matplotlib.pyplot as plt

# Same small training dataset for all three models
x_train = np.linspace(0, 6, 12)
noise = np.array([0.10, -0.18, 0.12, 0.06, -0.15, 0.18, -0.08, 0.12, -0.10, 0.08, -0.16, 0.10])
y_train = np.sin(x_train) + noise
x_smooth = np.linspace(0, 6, 400)
true_pattern = np.sin(x_smooth)

models = [
    (1, 'Underfitting', 'Too simple: misses the real curve', '#d62828'),
    (3, 'Good fit', 'Learns the useful pattern', '#2a9d8f'),
    (9, 'Overfitting', 'Chases noise in the training dots', '#6a4c93')
]

fig, axes = plt.subplots(1, 3, figsize=(15, 4.5), sharey=True)
for ax, (degree, title, message, color) in zip(axes, models):
    coefficients = np.polyfit(x_train, y_train, degree)
    prediction = np.polyval(coefficients, x_smooth)
    ax.scatter(x_train, y_train, color='#1d3557', s=38, label='Training data', zorder=3)
    ax.plot(x_smooth, true_pattern, '--', color='#8d99ae', linewidth=1.8, label='Underlying pattern')
    ax.plot(x_smooth, prediction, color=color, linewidth=2.5, label='Model')
    ax.set_title(title, color=color, weight='bold')
    ax.text(0.03, 0.04, message, transform=ax.transAxes, fontsize=9,
            bbox=dict(boxstyle='round,pad=0.35', facecolor='white', alpha=0.85))
    ax.set_xlabel('Input feature')
    ax.grid(alpha=0.2)
axes[0].set_ylabel('Target value')
axes[0].legend(loc='upper right', fontsize=8)
plt.suptitle('Choose the simplest model that still learns the real pattern', weight='bold', y=1.03)
plt.tight_layout()
plt.show()
```

## 6. Exam-ready recap

- Train data teaches the model; validation data helps choose the model; test data gives the final honest score.
- Underfitting: bad on train and test because the model is too simple.
- Overfitting: great on train but bad on unseen data because the model memorized noise.
- Generalization: good and similar performance on train and test data.

**One-line answer:** Overfitting happens when a model learns training details too closely and then performs poorly on new data.
