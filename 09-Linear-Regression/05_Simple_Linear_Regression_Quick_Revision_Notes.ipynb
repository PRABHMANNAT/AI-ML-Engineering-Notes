# Simple Linear Regression

## Quick, easy revision notes
Simple Linear Regression is a **supervised machine-learning algorithm** used to predict a number. It draws one best straight line through data.

## What problem does it solve?
Use it when you have **one input** and want to predict **one number**.

**Example: weight and height**
- Input $x$: a person's weight
- Output $y$: a person's height

We show the model old weight-height pairs. Then we give it a new weight and ask it to guess the height.

**Why is it called simple?** It has only one input feature. If we use many inputs, such as weight, age, and exercise time, it becomes Multiple Linear Regression.

## The main idea: draw the best line
Each dot is one old example. The model draws a line that stays as close as possible to all dots. This is called the **best-fit line**.

For a new input, move up from the x-axis until you touch the line. Then move left to read the predicted y-value.

```python
import numpy as np
import matplotlib.pyplot as plt

# Old examples: weight in kg and height in cm
weight = np.array([50, 55, 60, 65, 70, 75, 80])
height = np.array([155, 160, 164, 169, 171, 177, 181])

# Find one straight line: predicted_height = intercept + slope * weight
slope, intercept = np.polyfit(weight, height, 1)
predicted_height = intercept + slope * weight

# Ask the fitted line to predict height for a new weight
new_weight = 72
new_height = intercept + slope * new_weight

fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(weight, height, color='royalblue', s=70, label='Real old data')
ax.plot(weight, predicted_height, color='darkorange', linewidth=2.5, label='Best-fit line')

# These short vertical lines show the small mistakes (errors) made by the line
for x_value, real_y, line_y in zip(weight, height, predicted_height):
    ax.vlines(x_value, min(real_y, line_y), max(real_y, line_y), color='crimson', alpha=0.7)

ax.scatter(new_weight, new_height, color='green', marker='*', s=220, label='Prediction for new weight')
ax.annotate(f'Predicted height = {new_height:.1f} cm', (new_weight, new_height), xytext=(5, -28), textcoords='offset points')
ax.set_xlabel('Weight (kg)')
ax.set_ylabel('Height (cm)')
ax.set_title('Simple Linear Regression: dots, line, errors, and a prediction')
ax.grid(alpha=0.25)
ax.legend()
plt.show()

print(f'Line equation: predicted height = {intercept:.2f} + {slope:.2f} × weight')
print(f'For {new_weight} kg, predicted height is {new_height:.1f} cm')
```

### What the diagram and code mean
- Blue dots are real people from old data.
- Orange line is the model's best guess for the overall pattern.
- Red vertical lines are errors: the gap between a real dot and the line. Smaller gaps are better.
- Green star is the model's guess for a new weight.

The code first learns the line from old examples, then uses that line to predict one new height.

## The line equation
The model uses this equation:

$$ŷ = h_θ(x) = θ_0 + θ_1x$$

| Symbol | Easy meaning |
|---|---|
| $x$ | input, for example weight |
| $y$ | real answer from the data, for example real height |
| $ŷ$ or $h_θ(x)$ | model's predicted answer |
| $θ_0$ | intercept: where the line starts when $x = 0$ |
| $θ_1$ | slope: how much the prediction changes when $x$ goes up by 1 |

**Tiny example:** if $θ_0 = 120$ and $θ_1 = 0.8$, then for $x = 70$: $ŷ = 120 + 0.8(70) = 176$.

## Error and the best-fit line
For one dot, the error is:

$$error = y - ŷ$$

A good line has small errors for all dots. To stop positive and negative errors cancelling each other, we square them and average them. This is the **cost function**:

$$J(θ_0, θ_1) = (1 / 2m) ∑_{i=1}^{m} (ŷ^{(i)} - y^{(i)})^2$$

Do not worry about every symbol yet. It simply means: **square every mistake, add them, and make the total as small as possible.**

```python
# Try many possible slopes and see which one gives the smallest total squared error
possible_slopes = np.linspace(0.2, 1.4, 120)
costs = []

for trial_slope in possible_slopes:
    trial_intercept = np.mean(height - trial_slope * weight)
    trial_prediction = trial_intercept + trial_slope * weight
    costs.append(np.mean((height - trial_prediction) ** 2) / 2)

best_index = int(np.argmin(costs))
best_slope = possible_slopes[best_index]

plt.figure(figsize=(7, 4.5))
plt.plot(possible_slopes, costs, color='purple', linewidth=2.5)
plt.scatter(best_slope, costs[best_index], color='green', s=90, zorder=3, label='Smallest cost')
plt.xlabel('Possible slope values')
plt.ylabel('Cost (total squared mistake)')
plt.title('The best line sits at the bottom of the bowl')
plt.grid(alpha=0.25)
plt.legend()
plt.show()

print(f'Best slope found in this picture: {best_slope:.2f}')
```

### What the second diagram means
The purple curve is like a bowl. Every point on it is one possible line.

- High place on the bowl = a bad line with bigger mistakes.
- Bottom of the bowl = the best line with the smallest mistakes.

This is why the model changes $θ_0$ and $θ_1$: it wants to reach the bottom.

## Gradient Descent: how the model improves
Gradient Descent is a smart way to walk down the cost bowl without trying every possible line one by one.

1. Start with any line.
2. Check which direction makes the cost smaller.
3. Take a small step that way.
4. Repeat until the cost stops getting much smaller.

The small step size is called the **learning rate** ($α$). Too big can jump over the bottom; too small can take a very long time.

## One-minute revision
- Simple Linear Regression is supervised learning for predicting a number.
- It uses one input feature and one output.
- It finds a best-fit straight line.
- $θ_0$ is the intercept and $θ_1$ is the slope.
- $y$ is real; $ŷ$ is predicted.
- Error is the gap between real and predicted values.
- The best line has the smallest squared-error cost.
- Gradient Descent walks toward the bottom of the cost bowl.

**Memory trick:** `dots → line → errors → smallest cost → best prediction`.
